```python
import pandas as pd

vgg = pd.read_csv("pred_vgg_2.csv")
vit = pd.read_csv("pred_vit_2.csv")
rsmamba = pd.read_csv("pred_rsmamba_2.csv")

results = vgg.merge(
    vit,
    on=["chunk_id", "id_in_chunk", "flow"],
    how="inner"
)

results = results.merge(
    rsmamba,
    on=["chunk_id", "id_in_chunk", "flow"],
    how="inner"
)
```

```python
results["error_vgg"] = abs(results["flow"] - results["pred_vgg"])
results["error_vit"] = abs(results["flow"] - results["pred_vit"])
results["error_rsmamba"] = abs(results["flow"] - results["pred_rsmamba"])

results["max_error"] = results[
    ["error_vgg", "error_vit", "error_rsmamba"]
].max(axis=1)
```

```python
low = results[
    results["flow"] < 10
].nsmallest(4, "max_error")

medium = results[
    (results["flow"] >= 10) &
    (results["flow"] < 100)
].nsmallest(1, "max_error")

high = results[
    (results["flow"] >= 100) &
    (results["flow"] < 500)
].nsmallest(1, "max_error")

very_high = results[
    results["flow"] >= 500
].nsmallest(4, "max_error")

selected_indices = [
    low.iloc[3].name,
    medium.iloc[0].name,
    high.iloc[0].name,
    very_high.iloc[2].name
]

print(selected_indices)
```
